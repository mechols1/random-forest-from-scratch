```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from decision_tree import DecisionTree
```

```python
X, y = load_iris(return_X_y=True)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

## Decision Tree
This trains a single decision tree on the iris dataset and evaluatees accuracy

```python
tree = DecisionTree(max_depth=3)
tree.fit(X_train, y_train)
predictions = tree.predict(X_test)
accuracy = sum(predictions == y_test) / len(y_test)
print(f"Accuracy: {accuracy}")
tree.fit(X_train, y_train)
predictions = tree.predict(X_test)
```

```python
print("First 5 predictions:", predictions[:5])
print("First 5 actual:     ", list(y_test[:5]))
feature, threshold = tree._best_split(X_train, y_train)
print("Best feature:", feature)
print("Best threshold:", threshold)
```

```python
from random_forest import RandomForest

forest = RandomForest(tree_count=10)
forest.fit(X_train, y_train)
forest_predictions = forest.predict(X_test)
forest_accuracy = sum(forest_predictions == y_test) / len(y_test)
print(f"Decision Tree Accuracy: {accuracy}")
print(f"Random Forest Accuracy: {forest_accuracy}")
```

# Recall measurements : Decision Tree

```python
import numpy as np 
# decision tree 
recall_scores = {}
precision_scores = {}
for label in np.unique(y_test):
    true_positives = np.sum((predictions == label) & (y_test == label))
    false_negatives = np.sum((predictions != label) & (y_test == label))
    true_negative = np.sum((predictions != label) & (y_test != label))
    false_positives = np.sum((predictions == label) & (y_test != label))
    precision = true_positives / (true_positives + false_positives)
    recall = true_positives / (true_positives + false_negatives)
    recall_scores[label] = recall
    precision_scores[label] = precision
for key in recall_scores:
    print(f"Class {key} - Recall: {round(recall_scores[key], 2)}, Precision: {round(precision_scores[key], 2)}")
```
